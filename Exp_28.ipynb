```python
import re

print("=== Ethical Hacking Attack Simulation ===")

# Weak Password Analysis
password = input("\nEnter a password to analyze: ")

weak_passwords = ["123456", "password", "admin", "qwerty", "12345678"]

if password in weak_passwords or len(password) < 8:
    print("Weak Password Detected!")
else:
    print("Password Strength: Good")

# SQL Injection Detection
user_input = input("\nEnter SQL query input: ")

sql_patterns = [
    "' OR '1'='1",
    "' OR 1=1 --",
    "UNION SELECT",
    "DROP TABLE"
]

sql_detected = False

for pattern in sql_patterns:
    if pattern.lower() in user_input.lower():
        sql_detected = True
        break

if sql_detected:
    print("Potential SQL Injection Detected!")
else:
    print("No SQL Injection Pattern Found")

# Directory Traversal Detection
path_input = input("\nEnter file path request: ")

if "../" in path_input or "..\\" in path_input:
    print("Directory Traversal Attempt Detected!")
else:
    print("Path Appears Safe")

# Attack Logs
print("\n=== Attack Pattern Logs ===")

if password in weak_passwords or len(password) < 8:
    print("[LOG] Weak Password Found")

if sql_detected:
    print("[LOG] SQL Injection Pattern Detected")

if "../" in path_input or "..\\" in path_input:
    print("[LOG] Directory Traversal Attempt Detected")

# Vulnerability Report
print("\n=== Vulnerability Classification Report ===")

if password in weak_passwords or len(password) < 8:
    print("- Weak Authentication Vulnerability")

if sql_detected:
    print("- SQL Injection Risk")

if "../" in path_input or "..\\" in path_input:
    print("- Directory Traversal Risk")

# WAF Suggestions
print("\n=== WAF Rule Suggestions ===")
print("1. Block SQL keywords such as UNION SELECT and DROP TABLE")
print("2. Validate all user inputs")
print("3. Restrict '../' path access")
print("4. Enforce strong password policy")

print("\nSimulation Completed Successfully")
```
