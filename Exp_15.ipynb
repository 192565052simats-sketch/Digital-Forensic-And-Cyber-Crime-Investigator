```python
import hashlib

print("=== Malware Hash Analysis ===")

# Sample malware content
data = "This is a sample malware file for hash analysis"

# Generate hashes
md5_hash = hashlib.md5(data.encode()).hexdigest()
sha1_hash = hashlib.sha1(data.encode()).hexdigest()
sha256_hash = hashlib.sha256(data.encode()).hexdigest()

# Display hashes
print("\nGenerated Hashes:")
print("MD5    :", md5_hash)
print("SHA-1  :", sha1_hash)
print("SHA-256:", sha256_hash)

# Simple malware database
malware_database = {
    md5_hash: "Sample Malware"
}

# Detection
print("\nDetection Result:")
if md5_hash in malware_database:
    print("Malware Detected!")
    print("Malware Name:", malware_database[md5_hash])
else:
    print("No Malware Found")

# Detection Statistics
total_files = 1
detected_files = 1 if md5_hash in malware_database else 0

print("\nDetection Statistics:")
print("Total Files Scanned :", total_files)
print("Detected Malware    :", detected_files)

# Hash Collision Information
print("\nHash Collision Analysis:")
print("MD5 has higher collision risk.")
print("SHA-1 has moderate collision risk.")
print("SHA-256 is recommended for malware analysis.")

# Hash Family Clustering
print("\nHash Family Clustering:")
print("Family: Sample Malware")
print("MD5    :", md5_hash)
print("SHA-1  :", sha1_hash)
print("SHA-256:", sha256_hash)

print("\nExperiment Completed Successfully")
```
