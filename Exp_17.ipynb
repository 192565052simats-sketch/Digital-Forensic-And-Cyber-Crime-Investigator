```python
import psutil
import os
import tempfile

def capture_volatile_evidence():
    processes = [p.info for p in psutil.process_iter(['pid', 'name'])]
    return {
        "running_process_count": len(processes),
        "sample_processes": processes[:5],
    }

def capture_nonvolatile_evidence(folder):
    files = sorted(os.listdir(folder))
    return {"disk_files": files, "file_count": len(files)}

# Define evidence_dir. For demonstration, we'll use a temporary directory.
# In a real scenario, this would be a specific path you want to analyze.
with tempfile.TemporaryDirectory() as evidence_dir_path:
    # Create some dummy files in the temporary directory for testing
    with open(os.path.join(evidence_dir_path, "report.txt"), "w") as f:
        f.write("This is a report.")
    with open(os.path.join(evidence_dir_path, "log.csv"), "w") as f:
        f.write("id,value\n1,a\n2,b")

    volatile_snapshot = capture_volatile_evidence()
    nonvolatile_snapshot = capture_nonvolatile_evidence(evidence_dir_path)

    print("Volatile evidence -> running processes:",
          volatile_snapshot["running_process_count"])
    print("Non-volatile evidence -> disk files:", nonvolatile_snapshot)
```
